# Energy Calculator for wind and Solar


```python
def calculate_wind_energy(capacity_mw, capacity_factor_percent, hours):
  energy_mwh = capacity_mw * ((capacity_factor_percent / 100)) * hours
  return energy_mwh

# Given:
capacity_mw = 100
capacity_factor_percent = 35
hours = 24

result = calculate_wind_energy(capacity_mw, capacity_factor_percent, hours)
print(f"The total Wind Energy is {result} MWh")
```

```python
def calculate_solar_energy(capacity_mw, capacity_factor_percent,hours):
  energy_mwh = capacity_mw * ((capacity_factor_percent / 100)) * hours
  return energy_mwh

# Given:
capacity_mw = 50
capacity_factor_percent = 18
hours = 24

result = calculate_solar_energy(capacity_mw, capacity_factor_percent, hours)
print(f"The total solar Energy is {result} MWh")
```

## Docstring concepts to explain

- Directive (EU) 2023/2001 concerns renewable energy policy and builds on the EU Renewable Energy Directive framework.
- Wind and solar are recognised as renewable energy sources.
- Capacity factor is an analytical assumption/input in your calculator, not a single fixed capacity-factor value prescribed for every wind or solar project by the Directive.
- Actual capacity factors vary with location, technology, weather/resource conditions, curtailment, maintenance, and operational availability.
- For the EU accounting framework, wind generation has specific normalisation rules under Annex II
