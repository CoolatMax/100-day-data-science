- < €20/MWh

```python
renewable_capacity = {
    "Germany": 144.7,
    "France": 64.3,
    "Italy": 60.0,
    "Spain": 75.9,
    "Netherlands": 28.4,
    "Denmark": 10.2,
    "Portugal": 16.4,
    "Sweden": 33.1,
    "Poland": 22.7,
    "Belgium": 13.5
}
hourly_prices = {
    "Germany": [18, 25, 42, 85, 92, 67, 35, 12, 8, 22, 76, 105],
    "France": [15, 28, 45, 72, 88, 61, 30, 10, 19, 35, 82, 95],
    "Spain": [8, 12, 25, 55, 78, 91, 65, 32, 18, 5, 14, 48],
    "Denmark": [22, 35, 48, 95, 110, 72, 45, 18, 12, 30, 85, 102],
    "Netherlands": [19, 27, 51, 88, 96, 63, 41, 15, 9, 24, 79, 101]
}
prices = hourly_prices["Germany"]

# 1. Prices < €20/MWh
cheap_prices = [p for p in prices if p < 20]
print("Cheap (< 20):", cheap_prices)


# 2. Prices > €80/MWh
spike_prices = [p for p in prices if p > 80]
print("Spikes (> 80):", spike_prices)

# 3. Prices Between €20–€80/MWh (inclusive boundaries)
normal_prices = [p for p in prices if 20 <= p <= 80]
print("Normal (20-80):", normal_prices)
```

Count how many cheap and spike prices each country has

```python
renewable_capacity = {
    "Germany": 144.7,
    "France": 64.3,
    "Italy": 60.0,
    "Spain": 75.9,
    "Netherlands": 28.4,
    "Denmark": 10.2,
    "Portugal": 16.4,
    "Sweden": 33.1,
    "Poland": 22.7,
    "Belgium": 13.5
}
hourly_prices = {
    "Germany": [18, 25, 42, 85, 92, 67, 35, 12, 8, 22, 76, 105],
    "France": [15, 28, 45, 72, 88, 61, 30, 10, 19, 35, 82, 95],
    "Spain": [8, 12, 25, 55, 78, 91, 65, 32, 18, 5, 14, 48],
    "Denmark": [22, 35, 48, 95, 110, 72, 45, 18, 12, 30, 85, 102],
    "Netherlands": [19, 27, 51, 88, 96, 63, 41, 15, 9, 24, 79, 101]
}
for country, prices in hourly_prices.items():
    # Count prices less than 20
    cheap_count = len([p for p in prices if p < 20])

    # Count prices greater than 80
    spike_count = len([p for p in prices if p > 80])

    print(f"{country} -> Cheap Hours: {cheap_count}, Price Spikes: {spike_count}")
```

Calculate the average hourly price for each country.

```python
renewable_capacity = {
    "Germany": 144.7,
    "France": 64.3,
    "Italy": 60.0,
    "Spain": 75.9,
    "Netherlands": 28.4,
    "Denmark": 10.2,
    "Portugal": 16.4,
    "Sweden": 33.1,
    "Poland": 22.7,
    "Belgium": 13.5
}
hourly_prices = {
    "Germany": [18, 25, 42, 85, 92, 67, 35, 12, 8, 22, 76, 105],
    "France": [15, 28, 45, 72, 88, 61, 30, 10, 19, 35, 82, 95],
    "Spain": [8, 12, 25, 55, 78, 91, 65, 32, 18, 5, 14, 48],
    "Denmark": [22, 35, 48, 95, 110, 72, 45, 18, 12, 30, 85, 102],
    "Netherlands": [19, 27, 51, 88, 96, 63, 41, 15, 9, 24, 79, 101]
}
for country, prices in hourly_prices.items():
    # Calculate the average by dividing total sum by total number of hours
    average_price = sum(prices) / len(prices)

    # Print formatted to 2 decimal places
    print(f"{country}: €{average_price:.2f}/MWh")
```

Identify the country with the highest number of price spikes.

```python
renewable_capacity = {
    "Germany": 144.7,
    "France": 64.3,
    "Italy": 60.0,
    "Spain": 75.9,
    "Netherlands": 28.4,
    "Denmark": 10.2,
    "Portugal": 16.4,
    "Sweden": 33.1,
    "Poland": 22.7,
    "Belgium": 13.5
}
hourly_prices = {
    "Germany": [18, 25, 42, 85, 92, 67, 35, 12, 8, 22, 76, 105],
    "France": [15, 28, 45, 72, 88, 61, 30, 10, 19, 35, 82, 95],
    "Spain": [8, 12, 25, 55, 78, 91, 65, 32, 18, 5, 14, 48],
    "Denmark": [22, 35, 48, 95, 110, 72, 45, 18, 12, 30, 85, 102],
    "Netherlands": [19, 27, 51, 88, 96, 63, 41, 15, 9, 24, 79, 101]
}
# Use max() with a lambda key that filters and counts prices > 80 for each country
highest_spikes_country = max(
    hourly_prices,
    key=lambda country: len([p for p in hourly_prices[country] if p > 80])
)

# Calculate the actual count for the winning country
spike_count = len([p for p in hourly_prices[highest_spikes_country] if p > 80])

print(f"Country with the most price spikes: {highest_spikes_country} ({spike_count} spikes)")
```
