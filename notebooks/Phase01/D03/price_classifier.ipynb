## Write a function that takes hourly electricity prices and classifies them as 'cheap' (<€20/MWh), 'normal', or 'spike' (>€80/MWh).
### Loop through a list of prices.

```python
hourly_prices = [15, 22, 45, 85, 19, 67, 95, 35, 12, 81, 50, 78]
for price in hourly_prices:
  if price < 20:
    print(f"{price} is cheap")
  elif (price >= 20) & (price <= 80):
    print(f"{price} is normal")
  else:
    print(f"{price} is spike")
```
