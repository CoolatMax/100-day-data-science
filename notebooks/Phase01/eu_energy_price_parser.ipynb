## Main agenda of the problem:
Parse the CSV line by line
Extract:
- Date
- Country
- Electricity price
Clean the text using string methods
Convert each record into a structured dictionary
Store all records in a list
Generate a simple formatted report

### Problem 1: Read the file:

```python
# EU Electricity Price Parser
# Day 006 Mini Project

pricing_data = []

# Open the CSV file
with open("eu_pricing.csv", "r") as file:

    # Read the file line by line
    for line_number, line in enumerate(file):

        # Skip the header
        if line_number == 0:
            continue

        # Remove whitespace and newline
        line = line.strip()

        # Split the line into individual values
        data = line.split(",")

        # Extract values
        date = data[0].strip()
        country = data[1].strip()
        price = float(data[2].strip())

        # Classify electricity price
        if price < 20:
            classification = "cheap"

        elif price <= 80:
            classification = "normal"

        else:
            classification = "spike"

        # Create structured record
        record = {
            "date": date,
            "country": country,
            "price_eur_mwh": price,
            "classification": classification
        }

        # Store record
        pricing_data.append(record)


# Display parsed data
print("EU ELECTRICITY PRICE REPORT")
print("===========================")

for record in pricing_data:

    print(
        f"{record['date']} | "
        f"{record['country']:<10} | "
        f"€{record['price_eur_mwh']:.2f}/MWh | "
        f"{record['classification'].upper()}"
    )


# Summary calculations
cheap_count = 0
normal_count = 0
spike_count = 0
total_price = 0

for record in pricing_data:

    total_price += record["price_eur_mwh"]

    if record["classification"] == "cheap":
        cheap_count += 1

    elif record["classification"] == "normal":
        normal_count += 1

    else:
        spike_count += 1


# Calculate average price
average_price = total_price / len(pricing_data)


# Display summary
print("\nPRICE SUMMARY")
print("=============")

print(f"Total records: {len(pricing_data)}")
print(f"Cheap prices: {cheap_count}")
print(f"Normal prices: {normal_count}")
print(f"Price spikes: {spike_count}")
print(f"Average price: €{average_price:.2f}/MWh")
```
