## Problem 81

Calculate the solar plant's capacity factor and determine whether it is above 25%

```python
# Step 1: Define plant parameters (Sample values)
plant_capacity_mw = 10         # Total installed capacity in Megawatts
actual_generation_mwh = 22000  # Total energy produced over a year in Megawatt-hours
time_period_hours = 8760       # Total hours in a non-leap year (365 days * 24 hours)

# Step 2: Calculate the Capacity Factor
# Formula: (Actual Energy Generated) / (Max Possible Energy Generation if running at 100% capacity)
max_possible_generation = plant_capacity_mw * time_period_hours
capacity_factor = (actual_generation_mwh / max_possible_generation) * 100

# Step 3: Determine whether it is above 25% using a Boolean comparison
is_above_25 = capacity_factor > 25

# Step 4: Output the results
print(f"Calculated Capacity Factor: {capacity_factor:.2f}%")
print(f"Is it above 25%? {is_above_25}")
```
