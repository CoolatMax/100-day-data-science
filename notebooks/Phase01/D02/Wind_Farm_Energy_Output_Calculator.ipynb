```python
# Wind Farm Energy Output Calculator

# Inputs
capacity_mw = float(input("Enter wind farm capacity (MW): "))
capacity_factor = float(input("Enter capacity factor (%): "))
hours = float(input("Enter operating hours: "))

# Calculate energy output in MWh
energy_mwh = capacity_mw * (capacity_factor / 100) * hours

# Store results in a dictionary
wind_farm_results = {
    "capacity_mw": capacity_mw,
    "capacity_factor_percent": capacity_factor,
    "operating_hours": hours,
    "energy_output_mwh": energy_mwh
}

# Display results
print("\nWind Farm Results:")
for key, value in wind_farm_results.items():
    print(f"{key}: {value}")
```
